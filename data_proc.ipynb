```python
# function to calculate entropy for binary variables
def calculate_entropy(series):
    """Calculate the entropy of a binary pandas Series."""
    p = series.mean()
    if p == 0 or p == 1:
        return 0
    return -p * np.log2(p) - (1 - p) * np.log2(1 - p)
```

```python
ctr_aggregates = survey_data.groupby('alpha3').agg(
    belief_cc_mean =('belief_in_cc', 'mean'),
    belief_cc_std = ('belief_in_cc', 'std'),
    policy_support_mean = ('policy_support', 'mean'),
    policy_support_std = ('policy_support', 'std'),
    wept_mean = ('wept', 'mean'),
    wept_std = ('wept', 'std'),
    share_social_media_mean = ('share_sm', 'mean'),
    share_social_media_entropy = ('share_sm', calculate_entropy)
)
```

```python
ctr_aggregates
```

```python
def normalize_zero_hundred(series):
    """Normalize a pandas Series to a 0-100 scale."""
    min_val = series.min()
    max_val = series.max()
    return 100 * (series - min_val) / (max_val - min_val)
```

```python
for col in ctr_aggregates.columns:
    ctr_aggregates[col] = normalize_zero_hundred(ctr_aggregates[col])
```

```python
ctr_aggregates
```

```python
ctr_aggregates = ctr_aggregates.reset_index()
```
